```python
def clean_data(data):
    data = data.drop_duplicates()
    data = data.dropna(subset=['type', 'amount'])  
    data["type"] = data["type"].str.strip().str.upper()
    numeric_cols = data.select_dtypes(include=['number']).columns
    data = data[(data[numeric_cols] >= 0).all(axis=1)]
    
    return data.reset_index(drop=True)
```

```python
cleaned_data = clean_data(data)
print("Before: ", len(data), "rows")
print("After: ", len(cleaned_data), "rows")
```

```python
import pandas as pd

cols = ["type", "amount", "oldbalanceOrg", "newbalanceOrig",
        "oldbalanceDest", "newbalanceDest", "isFraud"]
df = pd.read_csv("cleaned.csv",usecols=cols)

fraud_types = df.loc[df["isFraud"] == 1, "type"].unique()
print("Types with fraud:", fraud_types)

df = df[df["type"].isin(fraud_types)].sample(200000, random_state=1)

df["errorOrig"] = df["newbalanceOrig"] + df["amount"] - df["oldbalanceOrg"]
df["errorDest"] = df["oldbalanceDest"] + df["amount"] - df["newbalanceDest"]
df["is_transfer"] = (df["type"] == "TRANSFER").astype(int)

features = ["amount", "oldbalanceOrg", "newbalanceOrig", "oldbalanceDest",
            "newbalanceDest", "errorOrig", "errorDest", "is_transfer"]
X, y = df[features], df["isFraud"]
print(y.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import joblib

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=1)

model = RandomForestClassifier(n_estimators=100, max_depth=12,
                               class_weight="balanced", n_jobs=-1, random_state=1)
model.fit(X_train, y_train)

print(classification_report(y_test, model.predict(X_test)))
joblib.dump(model, "model.pkl")
print("Saved model.pkl")
```
